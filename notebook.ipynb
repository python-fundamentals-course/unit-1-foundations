## 2. Variables — naming and assigning

A **variable** is a name that points to a value. You create one with the assignment operator `=`.

Python variable names:

- are case-sensitive (`age` and `Age` are different variables)
- use `snake_case` by convention (lowercase words separated by underscores)
- cannot start with a number or contain spaces


```python
course_name = "Python Fundamentals"
year = 2026

print(course_name)
print(year)
```

## 3. The Python data model — everything has a type

Python is **dynamically typed**: you don't declare a type up front, but every value still *has* a type. You can check it with the built-in `type()` function.

The four types you'll use most often in this course are:

| Type | Example | Meaning |
|------|---------|---------|
| `int` | `35` | a whole number |
| `float` | `19.99` | a number with a decimal point |
| `str` | `"Ada"` | text (a *string* of characters) |
| `bool` | `True` / `False` | a logical value |


```python
age = 35              # int
price = 19.99         # float
name = "Ada"          # str
is_enrolled = True    # bool

print(type(age))
print(type(price))
print(type(name))
print(type(is_enrolled))
```

### 3.1 `int` and `float` — numbers

Python supports the arithmetic operators you'd expect, plus a couple of extras:

- `/` always produces a `float` ("true division")
- `//` produces a whole number ("floor division")
- `%` gives the remainder ("modulo")
- `**` raises to a power


```python
a = 10
b = 3

print(a + b)    # addition
print(a - b)    # subtraction
print(a * b)    # multiplication
print(a / b)    # true division -> float
print(a // b)   # floor division -> int
print(a % b)    # modulo (remainder)
print(a ** b)   # exponentiation
```

### 3.2 `str` — text

Strings can be joined with `+` (concatenation), and Python's **f-strings** let you embed variables directly inside a string using `{ }`.


```python
first_name = "Jon"
last_name = "Scheele"
full_name = first_name + " " + last_name
print(full_name)

# An f-string: the f"" prefix lets you embed variables in {curly braces}
greeting = f"Hello, {first_name}! Welcome to {course_name}."
print(greeting)

print(full_name.upper())
print(full_name.lower())
print(len(full_name))
```

#### Working with parts of a string

A string is a *sequence* of characters, so two more tools let you work with pieces of it:

- **Indexing** — grab a single character by position using square brackets `[ ]`. Positions start at `0`, so `name[0]` is the *first* character. `name[-1]` counts back from the end.
- **`.split()`** — breaks a string into a list of words wherever there's a space (by default), so you can handle each word on its own.
- **`.title()`** — capitalises the first letter of each word.
- **`.replace(old, new)`** — replaces every occurrence of `old` with `new`.


```python
word = "lovelace"
print(word[0])        # 'l' - indexing starts at 0
print(word[-1])       # 'e' - negative indexes count back from the end

sentence = "ada lovelace"
words = sentence.split()     # splits on spaces -> ['ada', 'lovelace']
print(words)
print(words[0])        # 'ada'
print(words[0][0])     # 'a' - first character of the first word

print(sentence.title())            # 'Ada Lovelace' - capitalises each word
print(sentence.replace(" ", ""))   # 'adalovelace' - removes the spaces
```

### 3.3 `bool` — true or false

Booleans are the result of **comparisons** (`==`, `!=`, `<`, `>`, ...) and can be combined with `and`, `or`, and `not`.


```python
is_python_fun = True
has_completed_unit_1 = False

print(is_python_fun and has_completed_unit_1)
print(is_python_fun or has_completed_unit_1)
print(not has_completed_unit_1)

print(5 > 3)
print(5 == 3)
print(type(5 > 3))
```

## 4. `print()` in more detail

`print()` can take multiple comma-separated arguments — it joins them with a space and prints a newline at the end. F-strings also support **format specifiers**, such as `:.2f` to show a float with 2 decimal places.


```python
print("Name:", full_name, "| Age:", age)

print(f"{full_name} is {age} years old and the course price is ${price:.2f}.")
```
